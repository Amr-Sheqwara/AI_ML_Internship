## <span style="color: #39b64eff">2.1 Environment Setup & Library Imports</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

We begin by importing core numerical, data manipulation, visualization, and Scikit-Learn modeling and evaluation metrics.

</blockquote>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv("../../Data/raw/HousePriceDataset.csv")

print("✅ Libraries successfully imported, and dataset is loaded.")
```

## <span style="color: #309c42ff">2.2 Dataset Loading & Exploration</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

We load the housing dataset from <code>../../Data/raw/HousePriceDataset.csv</code> and inspect dataset dimensions, data types, and initial records.

</blockquote>

```python
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")

df.info()
```

## <span style="color: #309c42ff">2.3 Data Cleaning & Feature Selection</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

We remove non-numeric string columns (such as street address, city, statezip, country, and date) to keep numerical predictors suitable for linear regression modeling.

</blockquote>

```python
df['address'] = df['street'] + ", " + df['city'] + ", " + df['statezip'] + ", " + df['country']
df = df.drop(columns=['street', 'city', 'statezip', 'country'])
df = df.drop(columns=['address', 'date'])
```

## <span style="color: #309c42ff">2.4 Feature Matrix (x) & Target Vector (y) Separation</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

In Scikit-Learn conventions:
- <b>Feature Matrix ($X$):</b> A 2D DataFrame containing numerical predictor attributes.
- <b>Target Vector ($y$):</b> A 1D Series containing ground-truth price values (<code>price</code>).

</blockquote>

```python
target_col = "price"

x = df.drop(columns=[target_col])
y = df[target_col]

print(f"Feature Matrix (X) Shape: {x.shape}")
print(f"Target Vector (y) Shape: {y.shape}")
```

## <span style="color: #309c42ff">2.5 Linear Regression Model Training</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

We partition the dataset into 80% training data and 20% testing data (`random_state=42`), initialize `LinearRegression`, fit the model on the training set, and generate predictions for the test set.

</blockquote>

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

model = LinearRegression()
model.fit(x_train, y_train)
predictions = model.predict(x_test)

print("✅ Model Successfully Trained!")
```

## <span style="color: #309c42ff">2.6 Model Coefficients & Strongest Predictor</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

We extract model coefficients (`model.coef_`) to assess feature influence and sort them by absolute magnitude to identify the dominant predictor.

</blockquote>

```python
coefficients_df = pd.DataFrame({
    'Feature': x.columns,            
    'Coefficient': model.coef_,
    'Absolute_Effect': np.abs(model.coef_)
}).sort_values(by='Absolute_Effect', ascending=False)

print(coefficients_df)
```

## <span style="color: #309c42ff">2.7 Model Evaluation & Naive Baseline Comparison</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

We evaluate the model using MAE, RMSE, and $R^2$, and compare the model RMSE against a naive baseline predicting the average price (`y_train.mean()`).

</blockquote>

```python
mean_prediction = np.full_like(y_test, y_train.mean())
baseline_rmse = np.sqrt(mean_squared_error(y_test, mean_prediction))
rmse = np.sqrt(mean_squared_error(y_test, predictions))
mae = mean_absolute_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print("--- Baseline Performance ---")
print("Baseline RMSE: ", baseline_rmse)

print("\n--- Model Evaluation ---")
print("MAE (Mean Absolute Error): ",mae)
print("RMSE (Root Mean Squared Error): ",rmse)
print("R² (Coefficient of determination): ",r2)

if(rmse < baseline_rmse):
    result = baseline_rmse - rmse
    print("\n✅ Conclusion: The Linear Regression model ADDS VALUE over the mean baseline:\n ", result)
else:
    print("\n❌ Conclusion: The model DOES NOT add value over the baseline.")
```
