## Step 1: Import Libraries
Let's start by importing the core data science libraries we will use: Matplotlib, NumPy.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Simulate 10,000 coin flips with NumPy and confirm the proportion of heads approaches 0.5.
By simulating 10,000 coin flips, we are modeling a **Binomial distribution** which represents the number of successes in a fixed number of            
yes/no trials (where p = 0.5).

```python
np.random.seed(42)

flips = np.random.binomial(n = 1, p = 0.5, size = 10000)

proportion_heads = np.mean(flips)

print(f"Proportion of heads: {proportion_heads}")
```

## Step 2: Sample from a normal distribution with np.random.normal and plot its histogram to confirm the bell shape.
This visualizes a continuous **Normal (Gaussian) Distribution**. The histogram displays a symmetric bell curve shape, demonstrating that most randomly generated data points cluster tightly around the mean.

```python
np.random.seed(42)
normal_data = np.random.normal(loc=0, scale=1, size=10000)

plt.figure(figsize=(5, 3))
plt.hist(normal_data, bins=50, edgecolor='black')
plt.title("Histogram of a Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Step 3: Given a small worked scenario, compute a conditional probability by hand and verify it with a simulation.
This step demonstrates **conditional probability**: the likelihood of an event occurring given that another condition is already true.

### Computation by hand:

I want to calculate the probability of getting a 2 after throwing a dice if i knew the number shown will be Even.                         
[1 ,2 ,3 ,4 ,5 ,6]                                                                     
P(Even)= 3/6                                                   
P(2)=1/6                                               
P(2 | Even) = 1/6 / 3/6 = 1/3 = 0.33 #                                             

### Computation by simulation:

```python
Dice = [1, 2, 3, 4, 5, 6]
rolls = np.random.choice(Dice, size=100000)

even_rolls = rolls[rolls % 2 == 0]

Probability_of_2_given_Even = np.sum(even_rolls == 2) / len(even_rolls)

print(f"Theoretical P(2 | Even): ", 1/3)
print(f"Simulated P(2 | Even): ",  Probability_of_2_given_Even)
```
