## Step 1: Import Library
Let's start by importing the library we will use: NumPy.

## Step 2: Represent three data samples as a (3 × features) NumPy matrix

```python
x = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])
```

---
## Step 3: Compute the dot product of one sample vector with a weight vector by hand, then verify it with np.dot.


By hand: 

x = [2, 50000, 3]  w = [0.01, 0.25, 1.5]                         
x . w = (2 * 0.01) + (50000 * 0.25) + (3 * 1.5)                                             
x . w = 0.02 + 12500 + 4.5                               
x . w = 12504.52                         

Using np.dot

```python
w = np.array([0.01, 0.25, 1.5])
prediction = np.dot(x[0],w)
print("Dot product of one sample vector with a weight vector: ", prediction)
```

---
## Step 4: Use matrix multiplication (@) to produce a prediction for all three samples at once.

```python
predictions = x @ w 
print("prediction for all three samples at once", predictions)
```
