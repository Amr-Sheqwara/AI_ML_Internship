## Step 1: Import Libraries
Let's start by importing the core data science libraries we will use: NumPy, and Pandas.


```python
import pandas as pd
import numpy as np
```

---
## Step 2: Load and Inspect the Data
Now we will load the Titanic dataset (`tested.csv`) into a Pandas DataFrame and check its shape and structure.    
Then drop NaN values so they don't intervene with calculating the statistical formulas.

```python
df = pd.read_csv("../../Data/1.TitanicDataset.csv")
print("Dataset Info:")
df.info()
df = df.dropna(subset=["Age"])
```

---
## Step 3: Compute the mean, median, mode, standard deviation, and IQR.

```python
Age = df['Age']
print("Mean of the column Age =", np.mean(Age)) 
print("Median of the column Age =", np.median(Age)) 
print("Mode of the column Age =", Age.mode()[0]) 
print("Standard Deviation of the column Age =", np.std(Age))
print("Variance of the column Age =", np.var(Age)) 
q1, q3 = np.percentile(Age, [25, 75])
print("IQR of the column Age =", q3 - q1)
```


## Step 4:  Identify whether the mean or median better represents a 'typical' value, and justify the choice.
Using the median here is better to represent the typical age of people on the ship; because the bulk of the passengers are of age 18-27 but there's some Older people of Age 60-80 (Outliers) whom skew the dataset to the right
