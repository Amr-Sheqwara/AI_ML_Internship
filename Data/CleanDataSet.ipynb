```python
import pandas as pd
import numpy as np

# 1. Load the raw dataset
df = pd.read_csv("Data\penguins_size.csv")

# 2. Clean the dataset
df['sex'] = df['sex'].replace('.', np.nan)
df_clean = df.dropna(subset=['species', 'culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm', 'body_mass_g', 'sex']).copy()

# 3. Save the cleaned dataset to the Data folder (index=False avoids saving extra row index)
df_clean.to_csv("Data/penguins_cleaned.csv", index=False)

print(f"✅ Cleaned dataset saved successfully to 'Data/penguins_cleaned.csv'!")
print(f"Shape: {df_clean.shape[0]} rows, {df_clean.shape[1]} columns")
```
